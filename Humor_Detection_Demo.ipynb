```
# Download the fine-tuned MuRIL model (weights are stored with Git LFS)
SRC = 'https://media.githubusercontent.com/media/mahesh-2226/Hybrid-Humor-Detection-in-Code-Mixed-Tenglish-/main/muril_humor_model'
os.makedirs('muril_humor_model', exist_ok=True)
for f in ['config.json','tokenizer.json','tokenizer_config.json','model.safetensors']:
    if not os.path.exists(f'muril_humor_model/{f}'):
        !wget -q -O muril_humor_model/{f} {SRC}/{f}
!ls -lh muril_humor_model
```
